```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    "../dataset/Sample - Superstore.csv/Sample - Superstore.csv",
    encoding='latin1'
)

df.head()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales
```

```python
category_sales.plot(kind='bar')
plt.title('Category-wise Sales')
plt.ylabel('Sales')
plt.show()
```

```python
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
region_profit
```

```python
region_profit.plot(kind='bar')
plt.title('Region-wise Profit')
plt.ylabel('Profit')
plt.show()
```

```python
discount_profit = df.groupby('Discount')['Profit'].mean()

discount_profit.plot(kind='line')
plt.title('Discount vs Average Profit')
plt.ylabel('Average Profit')
plt.show()
```

```python
subcategory_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values()

subcategory_profit.plot(kind='barh')
plt.title('Sub-category Profitability')
plt.xlabel('Profit')
plt.show()
```
